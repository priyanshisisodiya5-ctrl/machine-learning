```python
from sklearn.preprocessing import PowerTransformer
# A right-skewed feature
X = np.array([
   [-1],
   [0],
   [3],
   [5],
   [10],
   [50]
])

# Create the transformer
pt = PowerTransformer(method='yeo-johnson')

 # Fit and transform
X_transformed = pt.fit_transform(X)
 
print('Original:')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))

print('\nLearned lamda:')
print(pt.lambdas_)
```

```python
from sklearn.preprocessing import PowerTransformer
# A right-skewed feature
X = np.array([
   [1],
   [1],
   [3],
   [5],
   [10],
   [50]
])

# Create the transformer
pt = PowerTransformer(method='box-cox')

 # Fit and transform
X_transformed = pt.fit_transform(X)
 
print('Original:')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))

print('\nLearned lamda:')
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

# Sample Dataset with Null values
data = {
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Vikram', 'Pooja', 'Rohan', 'Neha'],
    'Age': [25, np.nan, 30, 22, np.nan, 28, 35, 24],                             # Numerical (Null values)
    'Salary': [50000, 60000, np.nan, 45000, 80000, np.nan, 72000, 52000],        # Numerical (Null values)
    'Department': ['IT', 'HR', np.nan, 'Marketing', 'IT', 'Finance', np.nan, 'HR'],# Categorical (Null values)
    'City': ['Delhi', 'Mumbai', 'Delhi', np.nan, 'Bangalore', 'Mumbai', 'Delhi', np.nan], # Categorical (Null values)
    'Purchased': ['Yes', 'No', 'No', 'Yes', 'Yes', np.nan, 'No', 'Yes']          # Categorical (Null values)
}

df = pd.DataFrame(data)
print(df)
```
